# Multiple Linear Regression

### 원-핫 인코딩

```python
import pandas as pd
```

```python
dataset = pd.read_csv('MultipleLinearRegressionData.csv')
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:, -1].values
```

```python
X
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(drop='first'), [2])],remainder='passthrough')
X = ct.fit_transform(X)
X

# 1 0 : Home
# 0 1 : Library
# 0 0 : Cafe
```

### 데이터 세트 분리

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=0)
```

### 학습 (다중 선형 회귀)

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X_train, y_train)
```

### 예측 값과 실제 값 비교 (테스트 세트)

```python
y_pred = reg.predict(X_test)
y_pred
```

```python
y_test
```

```python
reg.coef_
```

```python
reg.intercept_
```

### 모델 평가

```python
reg.score(X_train, y_train) # 훈련 세트
```

```python
reg.score(X_test, y_test) # 테스트 세트
```

### 다양한 평가 지표 (회귀 모델)

1. MAE (Mean Absilute Error): 실제값과 예측값 차이의 절대값
2. MSE (Mean Squared Error): 실제값과 예측값 차이의 제곱
3. RMSE (Rote Mean Squared Error): 실제값과 예측값 차이의 제곱에 루트
4. R2 : 결정 계수 

> R2는 1에 가까울수록, 나머지는 0에 가까울수록 좋음

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test, y_pred) # 실제값, 예측값 #MAE
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test, y_pred) #MSE
```

```python
from sklearn.metrics import root_mean_squared_error
root_mean_squared_error(y_test, y_pred) #RMSE
```

```python
from sklearn.metrics import r2_score
r2_score(y_test, y_pred) #RMSE
```

# 단순선형회귀 실습 2 - 방 개수에 따른 집 가격 예측

- 독립 변수: 주택의 방 개수 (단위: 개) - 데이터: [1, 2, 3, 4, 5]
- 종속 변수: 집 가격 (단위: 억 원) - 데이터: [1.5, 2.8, 4.0, 5.5, 6.8]

###  실습 조건:
- 방이 3.5개(예: 방 3개 + 거실 겸 알파룸)인 집의 예상 가격을 예측해 보세요.
- 그래프에 한글 깨짐 현상을 해결하여 축 이름(Label)에 '방 개수'와 '집 가격(억 원)'을 정상적으로 출력하세요.

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = [[1],[2],[3],[4],[5]]
y = [1.5, 2.8, 4.0, 5.5, 6.8]
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
model.predict(X)
pred = model.predict([[3.5]])
pred
print('\n'+f"방이 3.5개인 집의 예상 가격: 약 {pred[0]:.2f}억 원")
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.scatter(3.5,pred,color='r',marker='^',s=100,label='예상 가격')
plt.scatter(X,y,color='b',label='실제 데이터')
plt.plot(X, model.predict(X),color='g',ls='--',label='회귀선')
plt.xlabel("방 개수")
plt.ylabel("집 가격(억 원)")
plt.legend()
```

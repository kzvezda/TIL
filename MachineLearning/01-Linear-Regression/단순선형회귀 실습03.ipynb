# 단순선형회귀 실습 3 - 자동차 주행거리와 중고차 가격 예측
1. 독립 변수 X (주행 거리, 단위: 만 km)
- X = [[1.2], [2.5], [4.0], [5.8], [7.2], [9.0], [11.5], [13.0]]
2. 종속 변수 y (중고차 가격, 단위: 만 원)
- y = [2800, 2450, 2100, 1700, 1450, 1100, 750, 400]


### 1. 데이터 분할 (Train / Test Split)

```python
X = [[1.2], [2.5], [4.0], [5.8], [7.2], [9.0], [11.5], [13.0]]
y = [2800, 2450, 2100, 1700, 1450, 1100, 750, 400]

from sklearn.model_selection import train_test_split 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

### 2. 모델 학습

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,y_train)
```

### 3. 새로운 데이터 예측

```python
model.predict(X_train)
pred = model.predict([[8]])
print(f"내 중고차가 8.0만 km를 달렸다면, 예상 가격은 {pred[0]:.2f}만 원입니다.")
```

### 4. 모델 평가

```python
model.score(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

### 5. 시각화 (차트 그리기)

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.scatter(X_train, y_train, color='black',label='학습 데이터')
plt.scatter(X_test, y_test, color='orange',label='시험 데이터')
plt.scatter(8, pred, color='g',marker='^',s=120)
plt.plot(X_train, model.predict(X_train),color='r',ls='-.',label='회귀선')
plt.xlabel('주행 거리 (만 km)')
plt.ylabel('중고차 가격 (만 원)')
plt.legend()
plt.show()
```
